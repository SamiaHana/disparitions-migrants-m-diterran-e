```python
import pandas as pd
import os

# Enregistrement dans un dataframe de la base de données
df = pd.read_csv("Missing_Migrants_Global_Figures_allData.csv")
```

```python
def map_to_continent(region_string):
    if pd.isna(region_string):
        return 'Unknown'
    
    region_string = region_string.lower().replace('(p)', '').strip()
    
    # Mapping simplifié
    if 'africa' in region_string:
        return 'Africa'
    elif 'asia' in region_string:
        return 'Asia'
    elif 'europe' in region_string:
        return 'Europe'
    elif 'oceania' in region_string:
        return 'Oceania'
    elif 'mixed' in region_string or 'unknown' in region_string:
        return 'Unknown'
    else:
        return 'Unknown'  # Catch-all

# Application sur une colonne Pandas
df['Continent of Origin'] = df['Region of Origin'].apply(map_to_continent)
```

```python
# On ne conserve que les incidents qui se sont produits en Europe ou en Méditerranée
df_filtre = df[df['Region of Incident'].isin(['Europe', 'Mediterranean'])]

# On ne garde que les informations qui nous intéressent (on conserve notamment les chiffres des disparitions mais ceux des décès)
colonnes_selectionnees = [
   'Region of Incident',
    'Country of Incident',
    'Migration Route',
    'Incident Date',
    'Incident Year',
    'Month',
    'Minimum Estimated Number of Missing', 
    'Continent of Origin'
]
df_filtre = df_filtre[colonnes_selectionnees]

# On enregistre le fichier final
df_filtre.to_csv("MissingMigrants_Europe_Med.csv", index=False)
```
